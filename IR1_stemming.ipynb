```python
import nltk
import re

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer

print("Libraries imported successfully!")
```

```python
# Download tokenizer and stopword datasets

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

print("NLTK datasets downloaded successfully!")
```

```python
# Sample text document

text = """
The students are studying information retrieval.
Information retrieval is an important part of web mining.
Students are learning how to search and retrieve documents.
Searching documents helps users find useful information quickly.
"""

print("ORIGINAL TEXT DOCUMENT")
print("----------------------")
print(text)
```

```python
# Convert all text into lowercase

lower_text = text.lower()

print("LOWERCASE TEXT")
print("--------------")
print(lower_text)
```

```python
# Remove punctuation and special characters

clean_text = re.sub(r'[^a-zA-Z\s]', '', lower_text)

print("CLEANED TEXT")
print("------------")
print(clean_text)
```

```python
# Split text into individual words

tokens = word_tokenize(clean_text)

print("TOKENIZED WORDS")
print("----------------")

print(tokens)

print("\nTotal number of tokens:", len(tokens))
```

```python
# Get English stop words
stop_words = set(stopwords.words('english'))

# Remove stop words from tokens
filtered_words = [
    word for word in tokens
    if word not in stop_words
]

print("WORDS AFTER STOP WORD REMOVAL")
print("-----------------------------")

print(filtered_words)

print("\nTotal words after stop word removal:",
      len(filtered_words))
```

```python
# Create Porter Stemmer object

stemmer = PorterStemmer()

# Apply stemming to filtered words

stemmed_words = [
    stemmer.stem(word)
    for word in filtered_words
]

print("WORDS AFTER STEMMING")
print("--------------------")

print(stemmed_words)
```

```python
# Convert stemmed words back into a sentence

final_text = ' '.join(stemmed_words)

print("FINAL PREPROCESSED DOCUMENT")
print("---------------------------")

print(final_text)
```

```python
print("TEXT PREPROCESSING COMPARISON")
print("=============================")

print("\nOriginal Text:")
print(text)

print("\nAfter Lowercase Conversion:")
print(lower_text)

print("\nAfter Tokenization:")
print(tokens)

print("\nAfter Stop Word Removal:")
print(filtered_words)

print("\nAfter Stemming:")
print(stemmed_words)

print("\nFinal Processed Text:")
print(final_text)
```

```python
import matplotlib.pyplot as plt

stages = [
    "Original Tokens",
    "After Stopword Removal",
    "After Stemming"
]

word_counts = [
    len(tokens),
    len(filtered_words),
    len(stemmed_words)
]

plt.figure(figsize=(9, 5))

plt.bar(stages, word_counts)

plt.xlabel("Preprocessing Stages")
plt.ylabel("Number of Words")
plt.title("Text Preprocessing Comparison")

plt.show()
```
